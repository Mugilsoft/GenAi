```python
import ollama
import gradio as gr

# System message (for context)
system_msg = "You are a tourist guide specializing in spiritual tourism in india."

# Chatbot function
def CustomLlama3(user_input):
    # Format the conversation context
    messages = [
        {"role": "system", "content": system_msg},
        {"role": "user", "content": user_input}
    ]

    # Generate response using Ollama's Llama 3.1 model
    response = ollama.chat(model="llama3.1", messages=messages)

    # Extract and return response text
    return response['message']['content']

# Gradio Interface
demo = gr.Interface(fn=CustomLlama3, 
                    inputs="text", 
                    outputs="text", 
                    title="My Travel Assistant")

# Launch the app
demo.launch(share=True)
```
