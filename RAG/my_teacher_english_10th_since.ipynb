```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma, FAISS
from langchain_text_splitters import RecursiveCharacterTextSplitter
import numpy
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("English.pdf")
docs = loader.load()
```

```python
# 1️⃣ Split your documents into smaller chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=900, chunk_overlap=200)
documents = text_splitter.split_documents(docs)
print(documents[:50])
```

```python
# 2️⃣ Create Hugging Face embeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
# 3️⃣ Create Chroma vector database
db = Chroma.from_documents(documents, embeddings)
```

```python
# 4️⃣ Perform similarity search
query = "இயந்திரவியல்"
retrieved_results = db.similarity_search(query)
print(retrieved_results[0].page_content)
```

```python
# 4️⃣ Perform similarity search
query = "A source producing a sound of frequency 90 Hz is approaching a stationary listener with a speed equal to (1/10) of the speed of sound. What will be the frequency heard by the listener?"
retrieved_results = db.similarity_search(query)
print(retrieved_results[0].page_content)
```

```python
# 4️⃣ Perform similarity search
query = "At what speed should a source of sound move away from a stationary observer so that observer finds the apparent frequency equal to half of the original frequency?"
retrieved_results = db.similarity_search(query)
print(retrieved_results[0].page_content)
```
