```python
import random
import csv

languages = [".NET", "Angular", "SQL"]

dotnet_templates = [
    ("Create a C# class for {entity}", 
     "public class {entity} {{ public int Id {{ get; set; }} public string Name {{ get; set; }} }}"),

    ("Create ASP.NET Core controller for {entity}",
     "[ApiController]\n[Route(\"api/[controller]\")]\npublic class {entity}Controller : ControllerBase {{ }}"),

    ("Create EF Core DbSet for {entity}",
     "public DbSet<{entity}> {entity}s {{ get; set; }}")
]

angular_templates = [
    ("Create Angular component for {entity}",
     "@Component({{ selector: 'app-{entity_lower}', templateUrl: './{entity_lower}.component.html' }})\nexport class {entity}Component {{ }}"),

    ("Create Angular service for {entity}",
     "@Injectable({{ providedIn: 'root' }})\nexport class {entity}Service {{ constructor(private http: HttpClient) {{ }} }}"),

    ("Create Angular form for {entity}",
     "this.form = this.fb.group({{ name: ['', Validators.required] }});")
]

sql_templates = [
    ("Write SQL query to get all {entity}",
     "SELECT * FROM {entity};"),

    ("Write SQL insert query for {entity}",
     "INSERT INTO {entity} (Name) VALUES ('John');"),

    ("Write SQL join query between {entity} and Orders",
     "SELECT * FROM {entity} e JOIN Orders o ON e.Id = o.{entity}Id;")
]

entities = ["Employee", "Product", "Order", "Customer", "Invoice", "Category","Sales","Purchase","Stock"]

dataset = []

for _ in range(1000):
    language = random.choice(languages)
    entity = random.choice(entities)

    if language == ".NET":
        template = random.choice(dotnet_templates)
    elif language == "Angular":
        template = random.choice(angular_templates)
    else:
        template = random.choice(sql_templates)

    instruction = template[0].format(entity=entity, entity_lower=entity.lower())
    response = template[1].format(entity=entity, entity_lower=entity.lower())

    dataset.append([instruction, response, language])

# Save CSV
with open("coding_dataset.csv", mode="w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["instruction", "response", "language"])
    writer.writerows(dataset)

print("✅ Dataset generated: coding_dataset.csv (1000 samples)")
```
